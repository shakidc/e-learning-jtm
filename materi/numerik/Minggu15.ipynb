```python
import numpy as np
```

```python
def forward_difference(f_next, f_curr, h):
    """Menghitung turunan pertama dengan metode Selisih Maju"""
    return (f_next - f_curr) / h

def central_difference_1st(f_next, f_prev, h):
    """Menghitung turunan pertama dengan metode Selisih Tengahan"""
    return (f_next - f_prev) / (2 * h)

def central_difference_2nd(f_next, f_curr, f_prev, h):
    """Menghitung turunan kedua dengan metode Selisih Tengahan"""
    return (f_next - 2 * f_curr + f_prev) / (h ** 2)
```

```python
# =====================================================================
# STUDI KASUS 1: Estimasi Kecepatan Aktuator Hidrolik (Selisih Maju)
# =====================================================================
print("=== STUDI KASUS 1 ===")
t_sk1 = np.array([0.2, 0.3, 0.4])
x_sk1 = np.array([0.050, 0.082, 0.120])
h_sk1 = t_sk1[1] - t_sk1[0]

# Evaluasi pada t = 0.2 (indeks 0), maka titik majunya adalah indeks 1
v_02 = forward_difference(x_sk1[1], x_sk1[0], h_sk1)
print(f"Kecepatan piston pada t = 0.2 s: {v_02:.2f} m/s\n")
```

```python
# =====================================================================
# STUDI KASUS 2: Gradien Temperatur pada Sirip Pendingin (Selisih Tengahan)
# =====================================================================
print("=== STUDI KASUS 2 ===")
x_sk2 = np.array([0.00, 0.05, 0.10])
T_sk2 = np.array([120.0, 98.5, 82.0])
h_sk2 = x_sk2[1] - x_sk2[0]

# Evaluasi pada x = 0.05 (indeks 1), titik belakang indeks 0, titik maju indeks 2
dTdx_005 = central_difference_1st(T_sk2[2], T_sk2[0], h_sk2)
print(f"Gradien temperatur pada x = 0.05 m: {dTdx_005:.1f} °C/m\n")
```

```python
# =====================================================================
# STUDI KASUS 3: Analisis Percepatan Vibrasi Bantalan Gelinding (Turunan Kedua)
# =====================================================================
print("=== STUDI KASUS 3 ===")
t_sk3 = np.array([0.04, 0.06, 0.08])
u_sk3 = np.array([2.15, 2.40, 2.22])
h_sk3 = t_sk3[1] - t_sk3[0]

# Evaluasi turunan kedua pada t = 0.06 (indeks 1)
a_006 = central_difference_2nd(u_sk3[2], u_sk3[1], u_sk3[0], h_sk3)
print(f"Percepatan vibrasi pada t = 0.06 s: {a_006:.1f} mm/s²")
print(f"Dalam satuan SI: {a_006/1000:.3f} m/s²\n")
```

```python
# =====================================================================
# STUDI KASUS 4: Menentukan Titik Puncak Kurva Efisiensi Turbin Gas
# =====================================================================
print("=== STUDI KASUS 4 ===")
rp_sk4 = np.array([8, 10, 12])
eta_sk4 = np.array([0.315, 0.340, 0.335])
h_sk4 = rp_sk4[1] - rp_sk4[0]

# Hitung turunan pertama dan kedua pada rp = 10 (indeks 1)
deta_drp = central_difference_1st(eta_sk4[2], eta_sk4[0], h_sk4)
d2eta_drp2 = central_difference_2nd(eta_sk4[2], eta_sk4[1], eta_sk4[0], h_sk4)

print(f"Turunan Pertama (Gradien) pada rp = 10 : {deta_drp:.4f}")
print(f"Turunan Kedua (Kelengkungan) pada rp = 10: {d2eta_drp2:.4f}")

# Logika pembuktian titik puncak
if abs(deta_drp) < 0.01 and d2eta_drp2 < 0:
    print("Kesimpulan: Terbukti rp = 10 adalah wilayah Efisiensi Termal Maksimum (Maksimum Lokal).")
elif abs(deta_drp) < 0.01 and d2eta_drp2 > 0:
    print("Kesimpulan: Terbukti rp = 10 adalah wilayah Minimum Lokal.")
else:
    print("Kesimpulan: Titik rp = 10 bukan merupakan titik puncak/ekstremum stasioner.")
```
